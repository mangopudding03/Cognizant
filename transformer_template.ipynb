# Transformer Text Generation

In this notebook, we will explore how transformer models (like GPT-2) can generate text based on a given prompt. We will experiment with generating text by adjusting parameters like temperature and sequence length.

## Instructions
1. Change the prompt below to experiment with different types of text generation.
2. Adjust the `max_length` and `temperature` parameters to see how they affect the output.
3. Generate at least 3 samples with different prompts and compare the results.


```python
from transformers import pipeline, GPT2Tokenizer

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token  # Set pad token to eos token

# Load GPT-2 text generation model
generator = pipeline('text-generation', model='gpt2')

# Set your prompt
prompt = 'In the future, education will'

# Generate text
result = generator(prompt, max_length=100, temperature=0.7, truncation = True, num_return_sequences=3, pad_token_id=tokenizer.eos_token_id)
print(result[0]['generated_text'])
```

```python
# News Headline
prompt = '\n The impact of AI on the future of work'
result = generator(prompt, max_length=50, temperature=0.7, truncation = True,  pad_token_id=tokenizer.eos_token_id)
print(result[0]['generated_text'] )
# Short story opener
prompt = '\n Once upon a time, there was a kingdom'
result = generator(prompt, max_length=100, temperature=0.7, truncation = True,  pad_token_id=tokenizer.eos_token_id)
print(result[0]['generated_text'])
#Dialogue
prompt = '\n Blessed be the fruit'
result = generator(prompt, max_length=100, temperature=0.7, truncation = True,  pad_token_id=tokenizer.eos_token_id)
print(result[0]['generated_text'])
```
